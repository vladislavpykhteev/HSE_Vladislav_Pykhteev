```python
class CourtCase:
    def __init__(self, case_number):
        # Обязательный номер судебного дела
        self.case_number = case_number

        # Атрибуты со значениями по умолчанию
        self.case_participants = []
        self.listening_datetimes = []
        self.is_finished = False
        self.verdict = ""

    def set_a_listening_datetime(self, listening_datetime):
        """Добавляет дату и время судебного заседания."""
        self.listening_datetimes.append(listening_datetime)

    def add_participant(self, participant):
        """Добавляет участника дела."""
        self.case_participants.append(participant)

    def remove_participant(self, participant):
        """Удаляет участника, если он присутствует в списке."""
        if participant in self.case_participants:
            self.case_participants.remove(participant)

    def make_a_decision(self, verdict):
        """Сохраняет решение и отмечает дело как завершённое."""
        self.verdict = verdict
        self.is_finished = True
```

```python
case = CourtCase("А60-12345/2026")

case.add_participant("6671000000")
case.add_participant("6672000000")

case.set_a_listening_datetime({
    "datetime": "25.09.2026 10:30",
    "courtroom": "Зал 5"
})

case.remove_participant("6672000000")

case.make_a_decision(
    "Исковые требования удовлетворить полностью"
)

print("Номер дела:", case.case_number)
print("Участники:", case.case_participants)
print("Заседания:", case.listening_datetimes)
print("Дело завершено:", case.is_finished)
print("Решение:", case.verdict)
```
